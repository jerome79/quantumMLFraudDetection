# Fraud Detection with Quantum Machine Learning

This notebook demonstrates the basic usage of the quantumml_fraud package for credit card fraud detection.

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from quantumml_fraud.data import DataLoader, DataPreprocessor
from quantumml_fraud.models import ClassicalMLModel, QuantumMLModel
from quantumml_fraud.evaluation import FraudDetectionMetrics
from quantumml_fraud.utils import Config

%matplotlib inline
```

## 1. Load Configuration

```python
config = Config('../configs/default_config.json')
print("Configuration loaded:")
print(config.to_dict())
```

## 2. Load and Explore Data

```python
# Load data
data_path = config.get('data.path')
print(f"Loading data from: {data_path}")

# Note: You need to download the dataset first
# Download from: https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud
# Place the creditcard.csv file in data/raw/
```

## 3. Data Preprocessing

```python
# Example with synthetic data
# Create synthetic data for demonstration
np.random.seed(42)
X_demo = np.random.randn(1000, 30)
y_demo = np.random.binomial(1, 0.1, 1000)  # 10% fraud rate

print(f"Demo data shape: {X_demo.shape}")
print(f"Fraud rate: {y_demo.mean():.2%}")
```

```python
# Preprocess data
preprocessor = DataPreprocessor(scaler_type='standard')
X_scaled = preprocessor.fit_transform(X_demo)

print(f"Scaled data - Mean: {X_scaled.mean():.4f}, Std: {X_scaled.std():.4f}")
```

## 4. Train Classical Model

```python
# Split data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_demo, test_size=0.2, random_state=42
)

print(f"Training set: {X_train.shape}")
print(f"Test set: {X_test.shape}")
```

```python
# Train classical model
classical_model = ClassicalMLModel(model_type='random_forest', n_estimators=100)
classical_model.train(X_train, y_train)

print("Classical model trained successfully!")
```

## 5. Evaluate Model

```python
# Make predictions
y_pred = classical_model.predict(X_test)
y_pred_proba = classical_model.predict_proba(X_test)

# Calculate metrics
metrics = FraudDetectionMetrics.calculate_all_metrics(y_test, y_pred, y_pred_proba)

print("Evaluation Metrics:")
for metric_name, metric_value in metrics.items():
    print(f"  {metric_name}: {metric_value:.4f}")
```

```python
# Confusion matrix
cm, breakdown = FraudDetectionMetrics.get_confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()

print("\nBreakdown:")
for key, value in breakdown.items():
    print(f"  {key}: {value}")
```

## 6. Quantum Model (Conceptual)

The quantum model implementation is a placeholder that demonstrates the structure.
Full quantum implementation would require quantum hardware access or simulators.

```python
# Initialize quantum model
quantum_model = QuantumMLModel(n_qubits=4, backend='qiskit')

# Train (placeholder implementation)
quantum_model.train(X_train[:100], y_train[:100], epochs=10)

print("Quantum model training demonstration completed!")
```
